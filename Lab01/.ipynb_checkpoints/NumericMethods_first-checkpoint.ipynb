```python
import numpy as np
import copy
import time
```

```python
def read_matrix_from_file(filename):
    with open(filename, 'r') as f:
        lines = f.readlines()
    n, m = map(int, lines[0].split())
    matrix = []
    for i in range(1, n + 1):
        row = list(map(float, lines[i].split()))
        matrix.append(row)
    return np.array(matrix)
```

## 1.1 LU-разложение с выбором опорного эл-та

```python
def lu_decomposition(A):
    n = A.shape[0]
    U = A.astype(float).copy()
    L = np.eye(n)
    P = np.eye(n)
    
    for k in range(n):
        max_row = k
        max_val = abs(U[k, k])
        for i in range(k + 1, n):
            if abs(U[i, k]) > max_val:
                max_val = abs(U[i, k])
                max_row = i
        
        if max_row != k:
            U[[k, max_row]] = U[[max_row, k]]
            P[[k, max_row]] = P[[max_row, k]]
            if k > 0:
                L[[k, max_row], :k] = L[[max_row, k], :k]
        
        if U[k, k] == 0:
            raise ValueError("Матрица вырожденная (нулевой элемент на диагонали)")
        
        for i in range(k + 1, n):
            L[i, k] = U[i, k] / U[k, k]
            for j in range(k, n):
                U[i, j] -= L[i, k] * U[k, j]
    
    return L, U, P


def lu_solve(L, U, P, b):
    n = len(b)
    
    b_new = P @ b.astype(float)
    
    y = np.zeros(n, dtype=float)
    for i in range(n):
        y[i] = b_new[i] - np.dot(L[i, :i], y[:i])
    
    x = np.zeros(n, dtype=float)
    for i in range(n - 1, -1, -1):
        x[i] = (y[i] - np.dot(U[i, i + 1:], x[i + 1:])) / U[i, i]
    
    return x


def lu_inverse(L, U, P):
    n = L.shape[0]
    A_inv = np.zeros((n, n), dtype=float)
    for i in range(n):
        e_i = np.zeros(n, dtype=float)
        e_i[i] = 1.0
        A_inv[:, i] = lu_solve(L, U, P, e_i)
    return A_inv

def lu_determinant(L, U, P):
    perm = np.argmax(P, axis=1)
    s = 0
    for i in range(P.shape[0]):
        if P[i, i] != 1:
            s += 1
    if s != 0: s -= 1
    
    det_P = (-1) ** s
    det_U = np.prod(np.diag(U))
    return det_P * det_U
```

```python
data = read_matrix_from_file('input_lu.txt')
A = data[:, :-1]  
b = data[:, -1]  

print(f"\nМатрица A ({A.shape[0]}x{A.shape[1]}):")
print(A)
print(f"\nВектор b: {b}")

# Само решение
L, U, P = lu_decomposition(A)

print("\n--- 1. Матрицы L и U ---")
print("\nМатрица L:")
print(np.round(L, 6))
print("\nМатрица U:")
print(np.round(U, 6))
print("\nМатрица перестановок P:")
print(P.astype(int))

# Проверка: LU = PA
print("\n--- 2. Результат умножения LU и проверка с PA ---")
LU_product = L @ U
print("LU =")
print(np.round(LU_product, 6))

print("\nPA =")
PA = P @ A
print(np.round(PA, 6))

# Решение системы
print("\n--- 3. Решение системы Ax = b ---")
x = lu_solve(L, U, P, b)
print(f"Решение x = {np.round(x, 6)}")
print(f"\nПроверка: Ax = {np.round(A @ x, 6)}")
print(f"Исходное b = {b}")

# Поиск обратной, ее проверка и детерминант
print("\n--- 4. Обратная матрица A^{-1} через LU-разложение ---")
A_inv = lu_inverse(L, U, P)
print("A^{-1} =")
print(np.round(A_inv, 6))

print("\n--- 5. Проверка A*A^{-1} = E ---")
AA_inv1 = A @ A_inv
AA_inv2 = A_inv @ A
print("A*A^{-1} =")
print(np.round(AA_inv1, 6))
print("A^{-1} * A =")
print(np.round(AA_inv2, 6))

print("\n--- 6. Определитель матрицы A через LU-разложение ---")
det_A = lu_determinant(L, U, P)
print(f"det(A) = {det_A}")

det_numpy = np.linalg.det(A)
print(f"det(A) (NumPy) = {det_numpy}")
```

## 1.2 Метод прогонки

```python
def read_tridiag_from_file(filename):
    with open(filename, 'r') as f:
        lines = f.readlines()
    n, m = map(int, lines[0].split())
    
    full = []
    for i in range(1, n + 1):
        row = list(map(float, lines[i].split()))
        full.append(row)
    
    full = np.array(full)
    
    rhs = full[:, -1]
    
    mat = full[:, :-1]
    
    main_diag = np.array([mat[i, i] for i in range(n)], dtype=float)
    
    lower_diag = np.zeros(n, dtype=float)
    for i in range(1, n):
        lower_diag[i] = mat[i, i - 1]
    
    upper_diag = np.zeros(n, dtype=float)
    for i in range(n - 1):
        upper_diag[i] = mat[i, i + 1]
    
    return main_diag, lower_diag, upper_diag, rhs
```

```python
def tridiag_solve(lower, main, upper, rhs):
    n = len(main)
    
    a = np.array(lower, dtype=float)   # нижняя диагональ
    b = np.array(main, dtype=float)    # главная диагональ
    c = np.array(upper, dtype=float)   # верхняя диагональ
    d = np.array(rhs, dtype=float)     # правая часть
    
    P = np.zeros(n-1, dtype=float) 
    Q = np.zeros(n-1, dtype=float)  
    
    # Прямой ход
    P[0] = -c[0] / b[0]
    Q[0] = d[0] / b[0]
    
    for i in range(1, n - 1):
        denom = b[i] + a[i] * P[i - 1]
        P[i] = -c[i] / denom
        Q[i] = (d[i] - a[i] * Q[i - 1]) / denom
    
    # Обратный ход
    x = np.zeros(n, dtype=float)
    x[n - 1] = (d[n - 1] - a[n - 1] * Q[n - 2]) / (b[n - 1] + a[n - 1] * P[n - 2])
    
    for i in range(n - 2, -1, -1):
        x[i] = P[i] * x[i + 1] + Q[i]
    
    return x, P, Q
```

```python
main_d, lower_d, upper_d, rhs = read_tridiag_from_file('input_tridiag.txt')
n = len(main_d)

print(f"\nВерхняя диагональ (c):  {upper_d}")
print(f"Главная диагональ (b):  {main_d}")
print(f"Нижняя диагональ (a):   {lower_d}")
print(f"Вектор правых частей:   {rhs}")

x, P_coeff, Q_coeff = tridiag_solve(lower_d, main_d, upper_d, rhs)

print(f"\n--- 1. Прогоночные коэффициенты ---")
print(f"P: {np.round(P_coeff, 6)}")
print(f"Q: {np.round(Q_coeff, 6)}")

print(f"\n--- 2. Решение системы ---")
print(f"x = {np.round(x, 6)}")

print(f"\nПроверка:")
A_tri = np.diag(main_d) + np.diag(lower_d[1:], -1) + np.diag(upper_d[:-1], 1)
print(f"Ax = {np.round(A_tri @ x, 6)}")
print(f"b  = {rhs}")
```

## Задача 1.3 — Метод простых итераций и метод Зейделя

```python
def simple_iteration(A, b, eps, max_iter=10000):
    n = len(b)
    x = np.zeros(n, dtype=float)
    history = []
    
    for k in range(max_iter):
        x_new = np.zeros(n, dtype=float)
        for i in range(n):
            s = 0.0
            for j in range(n):
                if j != i:
                    s += A[i, j] * x[j]
            x_new[i] = (b[i] - s) / A[i, i]
        
        diff = np.linalg.norm(x_new - x)
        history.append(diff)
        
        if diff < eps:
            return x_new, k + 1, history
        
        x = x_new.copy()
    
    print("Метод не сошёлся за максимальное число итераций")
    return x, max_iter, history


def seidel_method(A, b, eps, max_iter=10000):
    n = len(b)
    x = np.zeros(n, dtype=float)
    history = []
    
    for k in range(max_iter):
        x_new = np.copy(x)
        for i in range(n):
            s = 0.0
            for j in range(n):
                if j != i:
                    s += A[i, j] * x_new[j]  
            x_new[i] = (b[i] - s) / A[i, i]
        
        diff = np.linalg.norm(x_new - x)
        history.append(diff)
        
        if diff < eps:
            return x_new, k + 1, history
        
        x = x_new.copy()
    
    print("Метод Зейделя не сошёлся за максимальное число итераций")
    return x, max_iter, history
```

```python
data = read_matrix_from_file('input_iterative.txt')
A_iter = data[:, :-1]
b_iter = data[:, -1]
n_iter = A_iter.shape[0]

eps = 0.01

print(f"\nМатрица A ({n_iter}x{n_iter}):")
print(A_iter)
print(f"\nВектор b: {b_iter}")
print(f"\nЗаданная точность: epsilon = {eps}")

t0 = time.time()
x_jacobi, iter_jacobi, hist_jacobi = simple_iteration(A_iter, b_iter, eps)
t_jacobi = time.time() - t0

print("\n--- Метод простых итераций ---")
print(f"Количество итераций: {iter_jacobi}")
print(f"Время выполнения: {t_jacobi:.6f} с")
print(f"Решение: {np.round(x_jacobi, 10)}")


t0 = time.time()
x_seidel, iter_seidel, hist_seidel = seidel_method(A_iter, b_iter, eps)
t_seidel = time.time() - t0

print("\n--- Метод Зейделя ---")
print(f"Количество итераций: {iter_seidel}")
print(f"Время выполнения: {t_seidel:.6f} с")
print(f"Решение: {np.round(x_seidel, 10)}")


x_etallon = np.linalg.solve(A_iter, b_iter)

print("\n--- 3. Эталонное решение (NumPy) ---")
print(f"x_etallon = {np.round(x_etallon, 10)}")

print("\n--- Сравнение решений ---")
print(f"Отклонение Якоби от эталона:  {np.linalg.norm(x_jacobi - x_etallon):.2e}")
print(f"Отклонение Зейделя от эталона: {np.linalg.norm(x_seidel - x_etallon):.2e}")

print("\n--- 4. Сравнение методов по скорости сходимости ---")
print(f"Метод простых итераций: {iter_jacobi} итераций")
print(f"Метод Зейделя:          {iter_seidel} итераций")
ratio = t_jacobi / t_seidel
if ratio > 1:
    print(f"Метод Зейделя сходится быстрее примерно в {ratio:.1f} раз")
else:
    print(f"Метод простых итераций сходится быстрее примерно в {(1/ratio):.1f}")
```

## Задача 1.4 — Метод вращений (алгоритм Якоби) для собственных значений

```python
def jacobi_eigen(A, eps, max_iter=10000):
    n = A.shape[0]
    Ak = A.astype(float).copy()
    V = np.eye(n, dtype=float)
    
    for k in range(max_iter):
        max_val = 0.0
        p, q = 0, 1
        for i in range(n):
            for j in range(i + 1, n):
                if abs(Ak[i, j]) > max_val:
                    max_val = abs(Ak[i, j])
                    p, q = i, j
        
        # Проверка сходимости
        if max_val < eps:
            eigenvalues = np.diag(Ak)
            return eigenvalues, V, k
        
        if abs(Ak[p, p] - Ak[q, q]) < 1e-15:
            phi = np.pi / 4
        else:
            phi = 0.5 * np.arctan2(2.0 * Ak[p, q], Ak[p, p] - Ak[q, q])
        
        c = np.cos(phi)
        s = np.sin(phi)
        
        Ak_new = Ak.copy()
        
        for j in range(n):
            if j != p and j != q:
                Ak_new[p, j] = c * Ak[p, j] + s * Ak[q, j]
                Ak_new[q, j] = -s * Ak[p, j] + c * Ak[q, j]
                Ak_new[j, p] = Ak_new[p, j]
                Ak_new[j, q] = Ak_new[q, j]
        
        Ak_new[p, p] = c**2 * Ak[p, p] + 2 * s * c * Ak[p, q] + s**2 * Ak[q, q]
        Ak_new[q, q] = s**2 * Ak[p, p] - 2 * s * c * Ak[p, q] + c**2 * Ak[q, q]
        Ak_new[p, q] = 0.0
        Ak_new[q, p] = 0.0
        
        Ak = Ak_new
        
        V_new = V.copy()
        for i in range(n):
            V_new[i, p] = c * V[i, p] + s * V[i, q]
            V_new[i, q] = -s * V[i, p] + c * V[i, q]
        V = V_new
    
    print("Метод Якоби не сошёлся за максимальное число итераций")
    eigenvalues = np.diag(Ak)
    return eigenvalues, V, max_iter
```

```python
A_jac = read_matrix_from_file('input_jacobi.txt')
n_jac = A_jac.shape[0]

eps_jac = 0.01

print(f"\nМатрица A ({n_jac}x{n_jac}):")
print(A_jac)

eigenvalues, eigenvectors, num_iter = jacobi_eigen(A_jac, eps_jac)

print(f"\n--- 1. Заданная точность: epsilon = {eps_jac} ---")
print(f"\n--- 2. Найденные собственные значения ---")
print(f"Lambdas = {np.round(eigenvalues, 10)}")

print("\n--- 3. Собственные векторы и матрица собственных векторов ---")
for i in range(n_jac):
    print(f"\nλ_{i+1} = {eigenvalues[i]:.10f}")
    print(f"v_{i+1} = {np.round(eigenvectors[:, i], 10)}")

print(f"\nМатрица собственных векторов V:")
print(np.round(eigenvectors, 6))

print("\n--- 4. Проверка AV = VΛ ---")

Lambda = np.diag(eigenvalues)
AV = A_jac @ eigenvectors
VL = eigenvectors @ Lambda

print("AV =")
print(np.round(AV, 6))
print("\nVΛ =")
print(np.round(VL, 6))
```
